```python
from spire.doc import *
from spire.doc.common import *

# Create an object of Document class and load a Word document
doc = Document()
doc.LoadFromFile("guidelines.docx")

# Get the first section
section = doc.Sections.get_Item(0)

# Get the forth paragraph
paragraph = section.Paragraphs.get_Item(4)

# Add a comment to the paragraph
comment = paragraph.AppendComment("Lack of detail on the history of the currency.")

# Set the author of the comment
comment.Format.Author = "Steven Zambrano"

# Create a comment start mark and an end mark and set them as the start and end marks of the created comment
commentStart = CommentMark(doc, CommentMarkType.CommentStart)
commentEnd = CommentMark(doc, CommentMarkType.CommentEnd)
commentStart.CommentId = comment.Format.CommentId
commentEnd.CommentId = comment.Format.CommentId

# Insert the comment start mark and end mark at the beginning and end of the paragraph respectively
paragraph.ChildObjects.Insert(0, commentStart)
paragraph.ChildObjects.Add(commentEnd)

# Save the document
doc.SaveToFile("comments.docx")
doc.Close()
```

```python
from spire.doc import *
from spire.doc.common import *

# Create an object of Document class and load a Word document
doc = Document()
doc.LoadFromFile("guidelines.docx")

# Find the text to comment on
text = doc.FindString("permitted only with approval from the board", True, True)

# Create a comment and set the content and author of the comment
comment = Comment(doc)
comment.Body.AddParagraph().Text = "This is Katana rule ID."
comment.Format.Author = "Steven Zambrano"

# Get the found text as a text range and get the paragraph it belongs to
range = text.GetAsOneRange()
paragraph =  range.OwnerParagraph

# Add the comment to the paragraph
paragraph.ChildObjects.Insert(paragraph.ChildObjects.IndexOf(range) + 1, comment)

# Create a comment start mark and an end mark and set them as the start and end marks of the created comment
commentStart = CommentMark(doc, CommentMarkType.CommentStart)
commentEnd = CommentMark(doc, CommentMarkType.CommentEnd)
commentStart.CommentId = comment.Format.CommentId
commentEnd.CommentId = comment.Format.CommentId

# Insert the created comment start and end tags before and after the found text respectively
paragraph.ChildObjects.Insert(paragraph.ChildObjects.IndexOf(range), commentStart)
paragraph.ChildObjects.Insert(paragraph.ChildObjects.IndexOf(range) + 1, commentEnd)

# Save the document
doc.SaveToFile ("CommentOnText.docx")
doc.Close()
```

```python
import win32com.client as win32

try:
    # Open Word via COM automation
    word_app = win32.Dispatch("Word.Application")
    word_app.Visible = True  # Keep Word hidden while processing

    # Open the document
    doc1 = word_app.Documents.Open(r'C:\Users\steven\Desktop\Python\guidelines.docx')

    # Specify the target phrase and comment text
    target_phrase = "permitted"
    comment_text = "This paragraph mentions investment."
    author_name = "Steven Zambrano"

    # Loop through all paragraphs in the document
    for paragraph in doc1.Paragraphs:
        # Check if the paragraph contains the target phrase
        if target_phrase.lower() in paragraph.Range.Text.lower():
            # Add a comment to the paragraph
            paragraph.Range.Comments.Add(paragraph.Range, comment_text)

    # Save a copy of the document to avoid overwriting the original
    doc1.SaveAs2(r'C:\Users\steven\Desktop\Python\guidelinev2.docx')

    # Close the original document
    doc1.Close()

finally:
    # Ensure Word is properly closed
    word_app.Quit()

print(f"Comments added to paragraphs containing the phrase '{target_phrase}'")
```

This piece of code will allow you to search for specific phrases rather than loop through the entire document.ScribdDocument

import win32com.client as win32

try:
    # Open Word via COM automation
    word_app = win32.Dispatch("Word.Application")
    word_app.Visible = False  # Keep Word hidden while processing

    # Open the document
    doc = word_app.Documents.Open(r'C:\path\to\your_document.docx')

    # Specify the target phrase and comment text
    target_phrase = "investment"
    comment_text = "This paragraph mentions investment."
    author_name = "Steven Zambrano"

    # Use Word's built-in Find function to search for the phrase
    find = doc.Content.Find
    find.Text = target_phrase
    find.Forward = True  # Search forward in the document

    # Execute the search
    if find.Execute():  # If the phrase is found
        # Add a comment at the found location
        range_of_found_text = find.Parent  # This gives us the range of the found text
        range_of_found_text.Comments.Add(range_of_found_text, comment_text)

    # Save a copy of the document with the added comment
    doc.SaveAs2(r'C:\path\to\comments_with_investment_comment.docx')

    # Close the document
    doc.Close()

finally:
    # Ensure Word is properly closed
    word_app.Quit()

print(f"Comment added to the paragraph containing the phrase '{target_phrase}'")
